```python
def plot2(accion, nemo, color):
    HTMLcemargos = plotly_oneplot(accion, x_col="Fecha", y_col="Precios (último /cierre)", plot_title=f"{nemo} - Last Price: {accion["Precios (último /cierre)"].iloc[-1]:,.2f}", 
                                  color_in=color, date_marker = pd.to_datetime("2026-07-14"))
    grafico_html = HTMLcemargos.to_html(full_html=False, include_plotlyjs="cdn")
    input_table = company_table(datainput, nemo)
    tabla_html = input_table.to_html()
    graficoprueba = html_2_plot(grafico_html, tabla_html)
```
