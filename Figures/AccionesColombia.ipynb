```python
#Cargar Librerías
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
import os
import re
import openpyxl
```

```python
# Definimos una función para encontrar la raíz del proyecto
def project_root(marker=".git"):
    ruta = Path.cwd().resolve()
    while ruta != ruta.parent:
        if (ruta / marker).exists():
            return ruta
        ruta = ruta.parent
    raise FileNotFoundError(f"No se encontró '{marker}'")

PROJECT_DIR = project_root()
```

```python
# Leemos datos de acciones de Colombia desde un archivo CSV
data = pd.read_csv(f"{PROJECT_DIR}/Data/Colombia_Stocks/Acciones_colab.csv")
```

```python
data.tail(3)
```

```python
# Definimos la variable de Fecha en formato de fecha
data['Fecha'] = pd.to_datetime(data['Fecha'])
```

```python
data.info()
```

```python
Cemargos = data[data["Nemotécnico"] == "CEMARGOS"]
Cibest = data[data["Nemotécnico"] == "CIBEST"]
Ecopetrol = data[data["Nemotécnico"] == "ECOPETROL"]
GrupoAval = data[data["Nemotécnico"] == "GRUPOAVAL"]
Isa = data[data["Nemotécnico"] == "ISA"]
PFaval = data[data["Nemotécnico"] == "PFAVAL"]
PFDavigroup = data[data["Nemotécnico"] == "PFDAVIGRP"]
COLCAP = data[data["Nemotécnico"] == "ICOLCAP"]
```

```python
def grafica(dfinput, colores):

  # Graficar
  titulo = dfinput["Nemotécnico"].iloc[0]
  plt.figure(figsize=(12,6))
  plt.plot(dfinput["Fecha"], dfinput["Precios (último /cierre)"], color = colores, linewidth=2)

  plt.title(titulo)
  plt.xlabel("Fecha")
  plt.ylabel("Precio Cierre")
  plt.grid(True)

  plt.show()
```

```python
grafica(COLCAP, "black")
```

```python
grafica(Cemargos, "green")
```

```python
grafica(Cibest, "blue")
```

```python
grafica(Ecopetrol, "green")
```
